```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/raw/WA_Fn-UseC_-Telco-Customer-Churn.csv")
print(df.shape)
df.head()
```

```python
df.info()
```

```python
# pd.to_numeric with errors='coerce' turns non-numeric strings into NaN
blank_mask = pd.to_numeric(df["TotalCharges"], errors="coerce").isna()
print(f"Rows with non-numeric TotalCharges: {blank_mask.sum()}")
df[blank_mask][["customerID", "tenure", "TotalCharges", "MonthlyCharges"]]
```

```python
# Giving Unique values for every categoriacal columns
cat_cols = df.select_dtypes(include="object").columns.tolist()
cat_cols.remove("customerID")

for col in cat_cols:
    print(f"{col}: {df[col].unique()}")
```

```python
# Numeric column ranges
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")
df[["tenure", "MonthlyCharges", "TotalCharges"]].describe()
```

```python
# Target distribution (class imbalance)
df["Churn"].value_counts(normalize=True)

plt.figure(figsize=(5,4))
sns.countplot(data=df, x="Churn")
plt.title("Churn distribution")
plt.show()
```

```python
#  Key relationships with churn
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

sns.boxplot(data=df, x="Churn", y="tenure", ax=axes[0])
axes[0].set_title("Tenure vs Churn")

sns.countplot(data=df, x="Contract", hue="Churn", ax=axes[1])
axes[1].set_title("Contract type vs Churn")
axes[1].tick_params(axis='x', rotation=20)

sns.boxplot(data=df, x="Churn", y="MonthlyCharges", ax=axes[2])
axes[2].set_title("Monthly Charges vs Churn")

plt.tight_layout()
plt.show()
```

```python
# Correlation among numeric features
numeric_df = df[["tenure", "MonthlyCharges", "TotalCharges", "SeniorCitizen"]]
sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation matrix")
plt.show()
```
